1. Calculate Average Sales Amount by Region

```sql
SELECT 
    st.Name AS Territory,
    soh.TotalDue,
    AVG(soh.TotalDue) OVER (PARTITION BY st.Name) AS AvgPerTerritory
FROM Sales.SalesOrderHeader soh
JOIN Sales.SalesTerritory st ON soh.TerritoryID = st.TerritoryID;
```

2. Total Sales by Product Category per Year

```sql
SELECT *
FROM (
    SELECT 
        pc.Name AS Category,
        YEAR(soh.OrderDate) AS OrderYear,
        sod.LineTotal
    FROM Sales.SalesOrderHeader soh
    JOIN Sales.SalesOrderDetail sod ON soh.SalesOrderID = sod.SalesOrderID
    JOIN Production.Product p ON sod.ProductID = p.ProductID
    JOIN Production.ProductSubcategory psc ON p.ProductSubcategoryID = psc.ProductSubcategoryID
    JOIN Production.ProductCategory pc ON psc.ProductCategoryID = pc.ProductCategoryID
) AS SourceTable
PIVOT (
    SUM(LineTotal)
    FOR OrderYear IN ([2011], [2012], [2013], [2014])
) AS PivotTable;
```

3. How much earned per year, earned per item, and earned per item per year

```sql
SELECT 
    p.Name AS ProductName,
    YEAR(soh.OrderDate) AS OrderYear,
    SUM(sod.LineTotal) AS TotalSales
FROM Sales.SalesOrderDetail sod
JOIN Sales.SalesOrderHeader soh ON sod.SalesOrderID = soh.SalesOrderID
JOIN Production.Product p ON sod.ProductID = p.ProductID
GROUP BY GROUPING SETS (
    (p.Name, YEAR(soh.OrderDate)),
    (p.Name),
    (YEAR(soh.OrderDate)),
    ()
)
ORDER BY p.Name, OrderYear;
```

4. Running Total of Sales per Customer

```sql
SELECT 
    c.CustomerID,
    soh.OrderDate,
    soh.TotalDue,
    SUM(soh.TotalDue) OVER (PARTITION BY c.CustomerID ORDER BY soh.OrderDate) AS RunningTotal
FROM Sales.SalesOrderHeader soh
JOIN Sales.Customer c ON soh.CustomerID = c.CustomerID;
```

5. Compare Each Order to Previous for a Customer (good for tracking if theres an increase or decrease on spending per customer)

```sql
SELECT 
    c.CustomerID,
    soh.OrderDate,
    soh.TotalDue,
    LAG(soh.TotalDue) OVER (PARTITION BY c.CustomerID ORDER BY soh.OrderDate) AS PrevOrderTotal
FROM Sales.SalesOrderHeader soh
JOIN Sales.Customer c ON soh.CustomerID = c.CustomerID;
```

6. First Order for each customer

```sql
WITH RankedOrders AS (
    SELECT 
        soh.SalesOrderID,
        soh.CustomerID,
        soh.OrderDate,
        ROW_NUMBER() OVER (PARTITION BY soh.CustomerID ORDER BY soh.OrderDate) AS RowNum
    FROM Sales.SalesOrderHeader soh
)
SELECT *
FROM RankedOrders
WHERE RowNum = 1;
```

7. Divide Products into Quartiles Based on Total Sales

```sql
WITH ProductSales AS (
    SELECT 
        p.ProductID,
        p.Name,
        SUM(sod.LineTotal) AS TotalSales
    FROM Sales.SalesOrderDetail sod
    JOIN Production.Product p ON sod.ProductID = p.ProductID
    GROUP BY p.ProductID, p.Name
)
SELECT *,
    NTILE(4) OVER (ORDER BY TotalSales DESC) AS SalesQuartile
FROM ProductSales;
```

8. Rank Products by Sales Amount within each Category

```sql
SELECT 
    pc.Name AS Category,
    p.Name AS Product,
    SUM(sod.LineTotal) AS TotalSales,
    RANK() OVER (PARTITION BY pc.Name ORDER BY SUM(sod.LineTotal) DESC) AS SalesRank
FROM Sales.SalesOrderDetail sod
JOIN Production.Product p ON sod.ProductID = p.ProductID
JOIN Production.ProductSubcategory psc ON p.ProductSubcategoryID = psc.ProductSubcategoryID
JOIN Production.ProductCategory pc ON psc.ProductCategoryID = pc.ProductCategoryID
GROUP BY pc.Name, p.Name;
```

9\. Sales by year and month, and by only year

```sql
SELECT 
    YEAR(OrderDate) AS OrderYear,
    MONTH(OrderDate) AS OrderMonth,
    SUM(TotalDue) AS TotalSales
FROM Sales.SalesOrderHeader
GROUP BY GROUPING SETS (
    (YEAR(OrderDate), MONTH(OrderDate)),
    (YEAR(OrderDate)),
    ()
)
ORDER BY OrderYear, OrderMonth;
```

10. Days between current/most recent order and last order

```sql
SELECT 
    CustomerID,
    OrderDate,
    LAG(OrderDate) OVER (PARTITION BY CustomerID ORDER BY OrderDate) AS PrevOrderDate,
    DATEDIFF(DAY, LAG(OrderDate) OVER (PARTITION BY CustomerID ORDER BY OrderDate), OrderDate) AS DaysBetweenOrders
FROM Sales.SalesOrderHeader;
```
