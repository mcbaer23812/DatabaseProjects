```sql

USE WideWorldImporters;
GO
-- Query 1
-- Return all employees of the month for the year 2016. The employee of the month has the most orders that month.
SELECT OrderYear, [OrderMonth],  FullName AS EmployeeOfTheMonth, NumOrders 
FROM (
	SELECT 
		ROW_NUMBER() OVER (PARTITION BY YEAR(O.OrderDate), MONTH(O.OrderDate) ORDER BY COUNT(O.SalespersonPersonID) DESC) AS SalesRank,
		YEAR(O.OrderDate) AS OrderYear,
		DATENAME(MONTH, (O.OrderDate)) AS OrderMonth,
		P.FullName,
		COUNT(O.SalespersonPersonID) AS NumOrders
	FROM [Application].People AS P
		INNER JOIN Sales.[Orders] AS O
		ON P.PersonID = O.SalespersonPersonID
	WHERE Year(O.OrderDate)=2015
	GROUP BY YEAR(O.OrderDate), DATENAME(MONTH, (O.OrderDate)), MONTH(O.OrderDate), P.FullName
) AS RankedSales
WHERE SalesRank = 1
ORDER BY OrderYear, OrderMonth
```

```sql

USE WideWorldImporters;
GO
-- Query 2
-- Return all employees of the month for the year 2015. The employee of the month has the most orders that month.

WITH MonthlySales AS (
    SELECT 
        YEAR(O.OrderDate) AS OrderYear,
        DATENAME(MONTH, O.OrderDate) AS OrderMonth,
        MONTH(O.OrderDate) AS OrderMonthNumber,  
        P.FullName,
        COUNT(O.SalespersonPersonID) AS NumOrders
    FROM [Application].People AS P
    INNER JOIN Sales.[Orders] AS O
        ON P.PersonID = O.SalespersonPersonID
    WHERE YEAR(O.OrderDate) = 2015
    GROUP BY YEAR(O.OrderDate), DATENAME(MONTH, O.OrderDate), MONTH(O.OrderDate), P.FullName
),
RankedSales AS (
    SELECT 
        OrderYear,
        OrderMonth,
        OrderMonthNumber,
        FullName,
        NumOrders,
        ROW_NUMBER() OVER (PARTITION BY OrderYear, OrderMonthNumber ORDER BY NumOrders DESC) AS SalesRank
    FROM MonthlySales
)
SELECT 
    OrderYear, 
    OrderMonth, 
    FullName AS EmployeeOfTheMonth, 
    NumOrders
FROM RankedSales
WHERE SalesRank = 1
ORDER BY OrderYear, OrderMonthNumber; 
```

```sql
USE WideWorldImporters;
GO

-- Query 4
-- Give the email For the Contact person  of the supplier. If they have "NO LOGON", give the alternate contact person. 

SELECT 
    S.SupplierName,
    COALESCE(
        CASE 
            WHEN PC.IsPermittedToLogon = 0 THEN APC.EmailAddress  -- If "NO LOGON", use alternate contact
            ELSE PC.EmailAddress 
        END, 
        APC.EmailAddress  -- Fallback to alternate contact if primary has no email
    ) AS ContactEmail
FROM Purchasing.Suppliers AS S
LEFT JOIN Application.People AS PC  -- Primary contact
    ON S.PrimaryContactPersonID = PC.PersonID
LEFT JOIN Application.People AS APC  -- Alternate contact
    ON S.AlternateContactPersonID = APC.PersonID;
```

```sql
USE WideWorldImporters;
GO

-- Query 5
-- Retrieve a list of every order that has an employee attached and their phone number. 

SELECT 
    O.OrderID,
    O.OrderDate,
    O.ExpectedDeliveryDate,
    (
        SELECT P.PhoneNumber
        FROM Application.People AS P
        WHERE P.PersonID = O.SalespersonPersonID
    ) AS PhoneNumber,
    (
        SELECT P.FullName
        FROM Application.People AS P
        WHERE P.PersonID = O.SalespersonPersonID
    ) AS Salesperson
FROM Sales.Orders AS O
ORDER BY O.OrderDate;

```

```sql
USE WideWorldImporters;
GO

-- Query 7
-- Retrieve a list of the latest order for every sales person

SELECT 
    O.OrderID,
    O.OrderDate,
    O.ExpectedDeliveryDate,
    (
        SELECT MAX(O2.OrderDate)
        FROM Sales.Orders AS O2
        WHERE O2.SalespersonPersonID = O.SalespersonPersonID
    ) AS LatestOrderDate,
    (  
        SELECT P.FullName
        FROM Application.People AS P
        WHERE P.PersonID = O.SalespersonPersonID
    ) AS Salesperson
FROM Sales.Orders AS O
ORDER BY O.OrderDate;
```

```sql
USE WideWorldImporters;
GO

-- Query 8
-- Retrieve orders only with a total value greater than 10k from 2016


SELECT 
    O.OrderID,
    O.OrderDate,
    O.CustomerID,
    (SELECT SUM(OL.Quantity * OL.UnitPrice)  
     FROM Sales.OrderLines AS OL
     WHERE OL.OrderID = O.OrderID) AS TotalOrderValue
FROM Sales.Orders AS O
WHERE YEAR(O.OrderDate) = 2016  
  AND (SELECT SUM(OL.Quantity * OL.UnitPrice)  
       FROM Sales.OrderLines AS OL
       WHERE OL.OrderID = O.OrderID) > 10000  
ORDER BY O.OrderDate;
```

```sql
USE WideWorldImporters;
GO

-- Query 9
-- Return when a product is back ordered or undersupplied
SELECT 
    o.OrderID, 
    o.CustomerID, 
    o.SalespersonPersonID, 
    o.OrderDate, 
    o.ExpectedDeliveryDate, 
    o.BackorderOrderID, 
    (SELECT 
        CASE 
            WHEN o.BackorderOrderID IS NOT NULL THEN 'Backordered'
            WHEN o.IsUndersupplyBackordered = 1 THEN 'Undersupplied'
            ELSE 'Fulfilled'
        END
     ) AS OrderStatus
FROM Sales.Orders AS o;
```

```sql
USE WideWorldImporters;
GO

-- Query 10
-- Return the general location info of the customer and the name of the customer.
SELECT 
    c.CustomerID, 
    (SELECT REPLACE(c.CustomerName, 
                    SUBSTRING(c.CustomerName, CHARINDEX('(', c.CustomerName), LEN(c.CustomerName)), '') 
     ) AS CleanedCustomerName,
    (SELECT SUBSTRING(c.CustomerName, 
                     CHARINDEX('(', c.CustomerName) + 1, 
                     CHARINDEX(')', c.CustomerName) - CHARINDEX('(', c.CustomerName) - 1) 
     ) AS [Location]
FROM Sales.Customers AS c
WHERE CHARINDEX('(', c.CustomerName) > 0;
```
