# **PART \[I\] - SHOWCASE ALL 10 OF THE PROPOSITIONS AND QUERIES**

**PROPOSITION \[I\] - Orders with Customer Names**

- **Why Noteworthy?**
- _Provides a unified view of orders with customer identities, essential for customer service, invoicing, or analyzing purchasing patterns. Without this, orders would just be anonymous numbers!_

```sql
USE WideWorldImporters;
GO

SELECT O.OrderID AS "Identification Number(s) of the Order(s)", O.OrderDate AS "Date(s) of the Order(s)", C.CustomerName AS "Name(s) of the Customer(s)"
FROM Sales.Orders AS O
INNER JOIN Sales.Customers AS C ON O.CustomerID = C.CustomerID;
```

**PROPOSITION \[II\] - Orders With Above-Average Quantities**

- **Why Noteworthy?**
- _Identifies high-volume orders, which can highlight bulk buyers or popular products. This helps tailor inventory planning or promotional offers for high-value customers_

```sql
USE WideWorldImporters;
GO

SELECT OrderID AS "Identification Number(s) of the Order(s)", Quantity AS "Quantities"
FROM Sales.OrderLines
WHERE Quantity > (SELECT AVG(Quantity) FROM Sales.OrderLines);
```

**PROPOSITION \[III\] - Top Five Customers By Total Sales**

- **Why Noteworthy?**
- _Reveals your most valuable customers, enabling loyalty rewards, personalized marketing, or strategic partnerships. Knowing “who spends the most” drives revenue-focused decisions._

```sql
USE WideWorldImporters;
GO

WITH CustomerSales AS (SELECT C.CustomerName AS "Name(s) of the Customer(s)", SUM(OL.Quantity * OL.UnitPrice) AS "Total Expenditure" FROM Sales.Customers AS C INNER JOIN Sales.Orders AS O ON C.CustomerID = O.CustomerID INNER JOIN Sales.OrderLines AS OL ON O.OrderID = OL.OrderID GROUP BY C.CustomerName)
SELECT TOP 5 * FROM CustomerSales ORDER BY "Total Expenditure" DESC;
```

**PROPOSITION \[IV\] - Latest Order For Each Customer**

- **Why Noteworthy?**
- _Helps track customer engagement and retention. If a customer’s latest order is old, it might signal a need for re-engagement campaigns or feedback requests._

```sql
USE WideWorldImporters;
GO

SELECT CustomerID AS "Identification Number(s) of Customer(s)", OrderDate AS "Date(s) of the Order(s)"
FROM Sales.Orders AS O1
WHERE OrderDate = (SELECT MAX(OrderDate) FROM Sales.Orders AS O2 WHERE O1.CustomerID = O2.CustomerID);
```

**PROPOSITION \[V\] - Items And Their Suppliers**

- **Why Noteworthy?**
- _Critical for supply chain management. If a product is defective or popular, you can quickly identify its supplier to reorder stock or address issues._

```sql
USE WideWorldImporters;
GO

SELECT SI.StockItemName AS "Name(s) of Stock Item(s)", SI.SupplierID AS "Identification Number(s) of the Suppliers", S.SupplierName AS "Name(s) of the Suppliers"
FROM Warehouse.StockItems AS SI
INNER JOIN Purchasing.Suppliers AS S ON SI.SupplierID = S.SupplierID;
```

**PROPOSITION \[VI\] - Monthly Sales Versus Average**

- **Why Noteworthy?**
- _Highlights seasonal trends or anomalies. For example, a below-average month might require a marketing push, while a peak month could inform staffing or inventory needs._

```sql
USE WideWorldImporters;
GO

WITH MonthlySales AS (SELECT MONTH(OrderDate) AS Month, SUM(Quantity * UnitPrice) AS Sales FROM Sales.Orders AS O INNER JOIN Sales.OrderLines AS OL ON O.OrderID = OL.OrderID GROUP BY MONTH(OrderDate))
SELECT Month AS "Month(s)", Sales AS "Sale(s)", (SELECT AVG(Sales) FROM MonthlySales) AS "Yearly Average"
FROM MonthlySales;
```

**PROPOSITION \[VII\] -** **Employee Sales Performance**

- **Why Noteworthy?**
- _This analysis links employee performance to revenue, offering practical insights for managing the workforce. By measuring individual sales contributions, businesses can reward top performers to boost morale, identify those needing training, and strategically place high achievers in key roles. It allows the simplification of data analysis, allowing for quick comparisons over time or across regions._

```sql
USE WideWorldImporters;
GO

WITH EmployeeSales AS (SELECT P.FullName AS "Full Name(s)", SUM(OL.Quantity * OL.UnitPrice) AS "Total Sales" FROM Application.People AS P INNER JOIN Sales.Orders AS O ON P.PersonID = O.SalespersonPersonID INNER JOIN Sales.OrderLines AS OL ON O.OrderID = OL.OrderID GROUP BY P.FullName)
SELECT * FROM EmployeeSales ORDER BY "Total Sales" DESC;
```

**PROPOSITION \[VIII\] -** **Categorizing Sales by Size**

- **Why Noteworthy?**
- _Categorizing orders helps businesses segment customers and optimize inventory/resource allocation. For example, "Large" orders might indicate high-value clients needing priority support, while "Small" orders could highlight opportunities for upselling._

```sql
USE WideWorldImporters;
GO

WITH OrderSizes AS (SELECT OrderID AS "Identification Number(s) of the Order(s)", CASE WHEN SUM(Quantity * UnitPrice) < 1000 THEN 'Small' WHEN SUM(Quantity * UnitPrice) < 5000 THEN 'Medium' ELSE 'Large' END AS "Size(s) of Categories" FROM Sales.OrderLines GROUP BY OrderID)
SELECT * FROM OrderSizes;
```

**PROPOSITION \[IX\] -** **Average Order Value By Year**

- **Why Noteworthy?**
- _Tracking average order value (AOV) over time helps businesses measure customer spending habits and pricing strategy effectiveness. A rising AOV might indicate successful upselling or premium product adoption, while a decline could signal the need for promotions or customer retention efforts._

```sql
USE WideWorldImporters;
GO

SELECT Year AS "Year(s)", AVG(OrderTotal) AS "Average Order(s) Value(s)"
FROM (SELECT YEAR(OrderDate) AS Year, SUM(Quantity * UnitPrice) AS OrderTotal FROM Sales.Orders AS O INNER JOIN Sales.OrderLines AS OL ON O.OrderID = OL.OrderID GROUP BY YEAR(OrderDate), O.OrderID) AS YearlyOrders
GROUP BY Year;
```

**PROPOSITION \[X\] -** **Seasonal Sales Trends**

- **Why Noteworthy?**
- _This query is special because it transforms raw transaction data into actionable insights about seasonal patterns. Unlike static reports, it dynamically groups sales by time periods using simple but powerful date functions (`YEAR` and `MONTH`), making it easy to compare performance year-over-year or spot recurring trends._

```sql
USE WideWorldImporters;
GO

SELECT YEAR(OrderDate) AS "Year(s)", MONTH(OrderDate) AS "Month(s)", SUM(Quantity * UnitPrice) AS "Total Sale(s)"
FROM Sales.Orders AS O
JOIN Sales.OrderLines AS OL ON O.OrderID = OL.OrderID
GROUP BY YEAR(OrderDate), MONTH(OrderDate)
ORDER BY "Year(s)", "Month(s)";
```
