**Question 1:** <span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">Find customers who have placed orders but have not made any payments</span>

Related Tables: <span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">Sales.Customers, Sales.Orders, Sales.CustomerTransactions</span>

```sql
USE WideWorldImporters;

SELECT c.CustomerID, c.CustomerName
FROM Sales.Customers c -- aliasing; Sales.Customers c
INNER JOIN Sales.Orders o ON c.CustomerID = o.CustomerID
LEFT JOIN Sales.CustomerTransactions ct ON c.CustomerID = ct.CustomerID
WHERE ct.PaymentMethodID IS NULL;

/*
This query finds customers who have orders but no transactions by
performing a LEFT JOIN and filtering out those with null transactions. 
In Sales.CustomerTransactions table, PaymentMethodID is set to NULL if the payment is not cleared.
*/
```

2\. Retrieve all suppliers who have transactions exceeding their average transaction amount

Related Tables: Purchasing.Suppliers, Purchasing.SupplierTransactions

```sql
USE WideWorldImporters;

SELECT s.SupplierID, s.SupplierName, st.TransactionAmount
FROM Purchasing.Suppliers s
INNER JOIN Purchasing.SupplierTransactions st ON s.SupplierID = st.SupplierID
WHERE st.TransactionAmount > (
    SELECT AVG(TransactionAmount) FROM Purchasing.SupplierTransactions
);

/*
Explanation:
This query filters suppliers whose transaction amounts exceed the average by using a subquery to calculate the mean.
*/
```

3\. List the total number of orders each salesperson handled

Related Tables:

Sales.Orders, Application.People

```sql
Use WideWorldimporters;

SELECT p.PersonID, p.FullName, COUNT(o.OrderID) AS TotalOrders
FROM Application.People p
INNER JOIN Sales.Orders o ON p.PersonID = o.SalespersonPersonID
GROUP BY p.PersonID, p.FullName;

/*
Explanation:
This query joins People and Orders to count the number
 of orders each salesperson has processed.
*/
```

4\. Find the highest-priced order for each customer

Related Tables: Sales.Orders, Sales.OrderLines

```sql
Use WideWorldImporters;


SELECT o.CustomerID, MAX(ol.UnitPrice * ol.Quantity) AS MaxOrderPrice
FROM Sales.Orders o
INNER JOIN Sales.OrderLines ol ON o.OrderID = ol.OrderID
GROUP BY o.CustomerID;

/*
This query calculates the highest-priced order for each customer by joining Orders and OrderLines and finding the max value.
*/
```

5\. List all customers and their respective buying groups (if available)

Related Tables: Sales.Customers, Sales.BuyingGroups.

```sql
Use WideWorldImporters;

SELECT c.CustomerName, ISNULL(bg.BuyingGroupName, 'No Group') AS BuyingGroup
FROM Sales.Customers c
LEFT JOIN Sales.BuyingGroups bg ON c.BuyingGroupID = bg.BuyingGroupID;

/*
A LEFT JOIN ensures all customers are listed, including those without buying groups.
*/
```

6\. Find the most frequently used payment method

Related Tables: Application.PaymentMethods, Sales.CustomerTransactions.

```sql
Use WideWorldImporters;

SELECT pm.PaymentMethodName, COUNT(ct.PaymentMethodID) AS UsageCount
FROM Application.PaymentMethods pm
INNER JOIN Sales.CustomerTransactions ct ON pm.PaymentMethodID = ct.PaymentMethodID
GROUP BY pm.PaymentMethodName
ORDER BY UsageCount DESC;

/*
Groups transactions by payment method and counts occurrences.
*/
```

7\. List customers who have more than one contact person

Related Tables: Sales.Customers, Application.People.

```sql
Use WideWorldImporters;

SELECT c.CustomerID, c.CustomerName, COUNT(p.PersonID) AS ContactPersons
FROM Sales.Customers c
INNER JOIN Application.People p ON c.PrimaryContactPersonID = p.PersonID OR c.AlternateContactPersonID = p.PersonID
GROUP BY c.CustomerID, c.CustomerName
HAVING COUNT(p.PersonID) > 1;

/*
Uses OR conditions in JOIN to count contacts per customer and filters for multiple contacts.
*/
```

8\. List the most recent order for each customer

Related Tables: Sales.Orders, Sales.Customer.

```sql

Use WideWorldImporters;

SELECT o.CustomerID, c.CustomerName, o.OrderID, o.OrderDate
FROM Sales.Orders o
INNER JOIN Sales.Customers c ON o.CustomerID = c.CustomerID
WHERE o.OrderDate = (SELECT MAX(OrderDate) FROM Sales.Orders WHERE CustomerID = o.CustomerID);

-- This query retrieves the most recent order for each customer by using a subquery to 
-- find the maximum order date for each customer and then filtering orders accordingly.
```

9\. List customers who have placed at least 100 orders 

<span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">Related Tables: Sales.Customers, Sales.Orders</span>

```sql
Use WideWorldImporters;

SELECT c.CustomerID, c.CustomerName, COUNT(o.OrderID) AS TotalOrders
FROM Sales.Customers c
INNER JOIN Sales.Orders o ON c.CustomerID = o.CustomerID
GROUP BY c.CustomerID, c.CustomerName
HAVING COUNT(o.OrderID) >= 100;

/* This query groups customers based on their order count, filtering those who have placed at least three orders, which is useful for analyzing customer engagement. */
```

10\. Retrieve the total number of transactions and average <span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">transaction amount for each supplier&nbsp;</span>  

<span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">Related Tables: Purchasing.Suppliers, Purchasing.SupplierTransactions.</span>

```sql
Use WideWorldImporters;

SELECT s.SupplierID, s.SupplierName, COUNT(st.SupplierTransactionID) AS TotalTransactions, 
AVG(st.TransactionAmount) AS AvgTransactionAmount
FROM Purchasing.Suppliers s
INNER JOIN Purchasing.SupplierTransactions st ON s.SupplierID = st.SupplierID
GROUP BY s.SupplierID, s.SupplierName
ORDER BY TotalTransactions DESC;

/* This query aggregates transaction data for each supplier, 
showing the number of transactions and the average transaction amount 
to help in supplier evaluation and decision-making.  */
```
