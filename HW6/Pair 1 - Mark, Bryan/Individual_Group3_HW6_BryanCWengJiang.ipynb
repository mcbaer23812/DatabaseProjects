**Proposition I - <span style="color: var(--vscode-foreground);">Find Products That Have Been Both Sold To Customers And Purchased From Vendors</span>**

- **Explanation:** _This shows products that appear in both sales records and purchase records, indicating they’re part of the supply chain._

```sql
USE AdventureWorks2017;
GO

SELECT ProductID AS "Identification of Product(s)"
FROM Sales.SalesOrderDetail
INTERSECT
SELECT ProductID 
FROM Purchasing.PurchaseOrderDetail;
```

**Proposition II - <span style="color: var(--vscode-foreground);">Identify Employees Who Haven’t Created Any Sales Orders</span>**

- **Explanation:** _This lists employees not found in the sales order records, meaning they haven’t processed any sales._

```sql
USE AdventureWorks2017;
GO

SELECT CustomerID AS "Identification of Customer(s)"
FROM Sales.SalesOrderHeader  
WHERE YEAR(OrderDate) = 2013  
EXCEPT  
SELECT CustomerID  
FROM Sales.SalesOrderHeader  
WHERE YEAR(OrderDate) = 2014;  
```

**Proposition III -** **List Products That Have Neither Reviews Nor Sales**

- **Explanation:** _Shows products that are "invisible" to customers — never sold or reviewed._

```sql
USE AdventureWorks2017;
GO

SELECT ProductID AS "Identification of Product(s)"
FROM Production.Product
EXCEPT (SELECT ProductID FROM Production.ProductReview UNION SELECT ProductID FROM Sales.SalesOrderDetail);
```

**Proposition IV - <span style="color: var(--vscode-foreground);">Identify Vendors And Customers Sharing The Same City</span>**

- **Explanation:** _Shows cities where both vendors and customers are located, useful for logistics planning._

```sql
USE AdventureWorks2017;
GO

SELECT City
FROM Purchasing.Vendor
JOIN Person.BusinessEntityAddress ON Vendor.BusinessEntityID = BusinessEntityAddress.BusinessEntityID
JOIN Person.Address ON BusinessEntityAddress.AddressID = Address.AddressID
INTERSECT
SELECT City
FROM Sales.Customer
JOIN Person.BusinessEntityAddress ON Customer.PersonID = BusinessEntityAddress.BusinessEntityID
JOIN Person.Address ON BusinessEntityAddress.AddressID = Address.AddressID;
```

**Proposition V -** **<span style="color: var(--vscode-foreground);">List All Phone Numbers For Employees And Vendors (Including Duplicates)</span>**

- **Explanation:** _Combines phone numbers for employees and vendors, retaining duplicates._

```sql
USE AdventureWorks2017;
GO

SELECT PhoneNumber AS "Phone Number(s)"
FROM HumanResources.Employee 
INNER JOIN Person.PersonPhone ON Employee.BusinessEntityID = PersonPhone.BusinessEntityID
UNION ALL
SELECT PhoneNumber 
FROM Purchasing.Vendor 
INNER JOIN Person.PersonPhone ON Vendor.BusinessEntityID = PersonPhone.BusinessEntityID;
```

**Proposition VI - Create A List Of All Employee And Customer Names**

- **Explanation:** _Merges employee and customer names into one list, excluding duplicates._

```sql
USE AdventureWorks2017;
GO

SELECT FirstName AS "First Name(s)", LastName AS "Last Name(s)"
FROM HumanResources.Employee 
INNER JOIN Person.Person ON Employee.BusinessEntityID = Person.BusinessEntityID
UNION
SELECT FirstName, LastName 
FROM Sales.Customer 
INNER JOIN Person.Person ON Customer.PersonID = Person.BusinessEntityID;
```

**Proposition VII - <span style="color: var(--vscode-foreground);">Identify Products That Have </span> Never <span style="color: var(--vscode-foreground);">Been Reviewed </span> Or <span style="color: var(--vscode-foreground);">Sold</span>**

- **Explanation:** _Lists products missing reviews or sales (not necessarily both)._

```sql
USE AdventureWorks2017;
GO

(SELECT ProductID "Identification of Product(s)" FROM Production.Product EXCEPT SELECT ProductID FROM Production.ProductReview)
UNION
(SELECT ProductID "Identification of Product(s)" FROM Production.Product EXCEPT SELECT ProductID FROM Sales.SalesOrderDetail);
```

**Proposition VIII - <span style="color: var(--vscode-foreground);">Find Products With A Rating Of 4+ Stars </span> <span style="color: var(--vscode-foreground);">And</span> <span style="color: var(--vscode-foreground);"> Total Sales Exceeding 100 Units</span>**

- **Explanation:** _Identifies top-rated products that are also best-sellers._

```sql
USE AdventureWorks2017;
GO

SELECT ProductID  AS "Identification of Product(s)"
FROM Production.ProductReview 
WHERE Rating >= 4
INTERSECT
SELECT ProductID 
FROM Sales.SalesOrderDetail 
GROUP BY ProductID 
HAVING SUM(OrderQty) > 100;
```

**Proposition IX - <span style="color: var(--vscode-foreground);">Identify Customers Whose Order Totals Exceed The Average Order Value</span>**

- **Explanation:** _Targets high-value customers for loyalty programs._

```sql
USE AdventureWorks2017;
GO

SELECT CustomerID AS "Identification of Customer(s)"
FROM Sales.SalesOrderHeader 
WHERE TotalDue > (SELECT AVG(TotalDue) FROM Sales.SalesOrderHeader)
INTERSECT
SELECT CustomerID 
FROM Sales.Customer;
```

**Proposition X - <span style="color: var(--vscode-foreground);">Combine Historical Price Changes With Current Product Prices (Including Duplicates If Prices Were Unchanged)</span>**

- **Explanation:** _Creates a complete timeline of prices, including the current price (with `NULL` dates for clarity). Retains duplicates if a product’s current price matches its historical price._

```sql
USE AdventureWorks2017;
GO

SELECT ProductID AS "Identification of Product(s)", ListPrice AS "List of Price(s)", StartDate AS "Start of Date(s)", EndDate AS "End of Date(s)"
FROM Production.ProductListPriceHistory
UNION ALL
SELECT ProductID, ListPrice, NULL AS StartDate, NULL AS EndDate 
FROM Production.Product 
WHERE ListPrice IS NOT NULL;
```
