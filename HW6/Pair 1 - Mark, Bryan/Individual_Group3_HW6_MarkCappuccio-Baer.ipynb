## **1\. Combining Active and Inactive Customers** (UNION)

### **Why it’s useful:**

This query helps identify all customers in the system, both active and inactive, ensuring marketing efforts can reach everyone. We can also diversify marketing efforts, and tailor them differently towards active and inactive customers.

```sql
SELECT CustomerID, CustomerName, 'Active' AS Status
FROM Sales.Customers
WHERE IsOnCreditHold = 0
UNION
SELECT CustomerID, CustomerName, 'Inactive' AS Status
FROM Sales.Customers
WHERE IsOnCreditHold = 1;
```

## **2\. Listing All Stock Items and Their Suppliers** (UNION ALL)

### **Why it’s useful:**

Shows all stock items along with their suppliers, **including duplicate entries** where suppliers provide multiple stock items. This allows the company to see which suppliers hold which items.

```sql
SELECT StockItemID, StockItemName, 'Stock' AS Type
FROM Warehouse.StockItems
UNION ALL
SELECT SupplierID AS StockItemID, SupplierName AS StockItemName, 'Supplier' AS Type
FROM Purchasing.Suppliers;
```

## **3\. Identifying Common Delivery Cities for Customers and Suppliers** (INTERSECT)

### **Why it’s useful:**

This query WOULD find cities that serve both customers and suppliers if this database had that overlap, which in theory would allow businesses to optimize delivery logistics. If suppliers and customers are in the same cities, you could rely more on those suppliers in the city for faster delivery to those customers. For example, I wouldn't prefer going to the supplier based in rural Missouri for my customers in New York.

```sql
SELECT DeliveryCityID
FROM Sales.Customers
INTERSECT
SELECT DeliveryCityID
FROM Purchasing.Suppliers;
```

## **4\. Finding Common Products Between Stock and Orders** (INTERSECT)

### **Why it’s useful:**

Ensures that all products in stock are actually being ordered, helping businesses assess slow moving inventory.

```sql
SELECT StockItemID FROM Warehouse.StockItems
INTERSECT
SELECT StockItemID FROM Sales.OrderLines;
```

## **5\. Customers Who Have Never Placed an Order** (EXCEPT)

### **Why it’s useful:**

Identifies customers who have never made a purchase, enabling businesses to target them with specific promotions. On a large scale, a company might see trends in customer account creation location and order placed volume.

```sql
SELECT CustomerID, CustomerName 
FROM Sales.Customers
EXCEPT
SELECT DISTINCT CustomerID, NULL 
FROM Sales.Orders;
```

## **6\. Products That Have Never Been Ordered** (EXCEPT)

### **Why it’s useful:**

Finds stock items that were never purchased, helping in inventory management. You could order more or less products based on if none were ordered. You could also change this to be some kind of threshold instead.

```sql
SELECT StockItemID, StockItemName 
FROM Warehouse.StockItems
EXCEPT
SELECT DISTINCT StockItemID, NULL 
FROM Sales.OrderLines;
```

## **7\. Top Selling Employees vs Employees Who Haven't Sold** (EXCEPT)

### **Why it’s useful:**

Finds employees who have never processed sales, which helps in performance reviews. This database has no salespeople who haven't sold since it is small, but I thought this query would be helpful nonetheless. You could also modify this to be based on a threshold -- a certain number of sales.

```sql
SELECT p.PersonID, p.FullName 
FROM Application.People p
WHERE p.IsSalesperson = 1  -- Only include actual salespeople
EXCEPT
SELECT DISTINCT o.SalespersonPersonID, p.FullName
FROM Sales.Orders o
JOIN Application.People p ON o.SalespersonPersonID = p.PersonID;
```

## **8\. Listing Distinct Customer and Supplier Locations** (UNION)

### **Why it’s useful:**

Combines all locations where customers and suppliers operate for business expansion insights.

```sql
SELECT DeliveryCityID FROM Sales.Customers
UNION
SELECT DeliveryCityID FROM Purchasing.Suppliers;
```

## **9\. Combining Customer and Supplier Contact Information** (UNION ALL)

### **Why it’s useful:**

Creates a contact list with both customers and suppliers, allowing a company to assess all phone numbers they need to include in marketing and other forms of contact.

```sql
SELECT CustomerName AS Name, PhoneNumber FROM Sales.Customers
UNION ALL
SELECT SupplierName AS Name, PhoneNumber FROM Purchasing.Suppliers;
```

## **10\. Identifying Locations That Are Only Supplier-Based** (EXCEPT)

### **Why it’s useful:**

Finds supplier locations that have no customers, helping in judging product demand. Cities with no customers will have a lot less local demand, implying less of a need for product. A company could take a look at this list and analyze the reason for the supplier.

```sql
SELECT DeliveryCityID FROM Purchasing.Suppliers
EXCEPT
SELECT DeliveryCityID FROM Sales.Customers;
```
