<span style="font-size: 10pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span>

```sql
USE AdventureWorks2019;

SELECT DISTINCT JobTitle AS TitleOrDepartment
FROM HumanResources.Employee

UNION

SELECT DISTINCT Name AS TitleOrDepartment
FROM HumanResources.Department;
```

Find employees who worked in multiple departments</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Related Tables: HumanResources.EmployeeDepartmentHistory</span><span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span><span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Why useful: Helps HR track employees who have worked across different departments.</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span>

```sql
USE AdventureWorks2019;
SELECT BusinessEntityID FROM HumanResources.EmployeeDepartmentHistory
INTERSECT
SELECT BusinessEntityID FROM HumanResources.EmployeeDepartmentHistory
GROUP BY BusinessEntityID
HAVING COUNT(DISTINCT DepartmentID) > 1;
/*This query finds employees appearing more than once in department history, indicating they worked in multiple departments.*/
```

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Related Tables: HumanResources.EmployeePayHistory, HumanResources.EmployeeDepartmentHistory</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span>

```sql
USE AdventureWorks2019;
SELECT BusinessEntityID FROM HumanResources.EmployeePayHistory
GROUP BY BusinessEntityID
HAVING COUNT(*) > 1
EXCEPT
SELECT BusinessEntityID FROM HumanResources.EmployeeDepartmentHistory
GROUP BY BusinessEntityID
HAVING COUNT(DISTINCT DepartmentID) > 1;

--Explanation: Retrieves employees with multiple pay changes but no department switch

```

Find Products That Are Both Purchased and Sold</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span>

<span style="font-size: 14pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-weight: 700; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">8. Find the products that have been ordered more than once.</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Related Tables: Purchasing.PurchaseOrderDetail</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Why useful: This query helps in identifying popular products that are frequently ordered, which can be useful for inventory planning.</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span>

```sql
USE AdventureWorks2019;
SELECT ProductID
FROM Purchasing.PurchaseOrderDetail
GROUP BY ProductID
HAVING COUNT(ProductID) > 1
INTERSECT
SELECT ProductID
FROM Purchasing.PurchaseOrderDetail;
/*Explanation: This query uses the INTERSECT operator to find products that appear more than once in the Purchasing.PurchaseOrderDetail table.*/
```

<span style="font-size: 14pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-weight: 700; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">9. Retrieve the list of products that have never been reviewed.</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Related Tables: Production.Product, Production.ProductReview</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Why useful: This query is useful for identifying products that may need more customer feedback or reviews</span>

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;"><br></span>

```sql
USE AdventureWorks2019;
SELECT ProductID, Name
FROM Production.Product
EXCEPT
SELECT ProductID, Name
FROM Production.Product
WHERE ProductID IN (SELECT ProductID FROM Production.ProductReview);
/*
Explanation: This query uses the EXCEPT operator to find products that have no reviews in the Production.ProductReview table.
*/
```

<span style="font-size: 13pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-weight: 700; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">10. Find all products that have been reviewed, and ordered, and have a bill of materials.</span>

  

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Related Tables: Production.Product, Production.ProductReview, Purchasing.PurchaseOrderDetail, Production.BillOfMaterials</span>

  

<span style="font-size: 11pt; font-family: Arial, sans-serif; color: rgb(0, 0, 0); background-color: transparent; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">Why useful: To identify popular products that are both manufactured internally and actively sold, which can be crucial for product lifecycle management and capacity planning.</span>

```sql
USE AdventureWorks2019;
SELECT p.ProductID, p.Name
FROM Production.Product p
INNER JOIN Production.ProductReview pr ON p.ProductID = pr.ProductID
INTERSECT
SELECT p.ProductID, p.Name
FROM Production.Product p
INNER JOIN Purchasing.PurchaseOrderDetail pod ON p.ProductID = pod.ProductID
INTERSECT
SELECT p.ProductID, p.Name
FROM Production.Product p
INNER JOIN Production.BillOfMaterials bom ON p.ProductID = bom.ComponentID
ORDER BY ProductID;

/*
This query finds products that exist in all three categories: reviewed, ordered, and having a bill of materials. It uses the INTERSECT operator twice to find the intersection of these three sets, identifying products that are both popular with customers and manufactured internally.
*/
```
